```python
import pandas as pd
```

```python
# Read datasets
fake_data = pd.read_csv("/content/Fake.csv")
true_data = pd.read_csv("/content/True.csv")

# Show first 5 rows
fake_data.head()
```

```python
print("Fake News Rows:", len(fake_data))
print("True News Rows:", len(true_data))
```

```python
# Add labels
fake_data["label"] = "Fake"
true_data["label"] = "Real"

print(fake_data.head())
```

```python
# Combine both datasets
data = pd.concat([fake_data, true_data])

# Shuffle data
data = data.sample(frac=1)

# Show first 5 rows
data.head()
```

```python
# Keep only text and label columns
data = data[["text", "label"]]

# Show dataset
data.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
```

```python
# Convert text into numbers
vectorizer = CountVectorizer()

x = vectorizer.fit_transform(data["text"])

y = data["label"]
```

```python
# Split data
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

print("Data split completed!")
```

```python
# Train model
model = MultinomialNB()

model.fit(x_train, y_train)

print("Model trained successfully!")
```

```python
# Test news
news = ["Breaking: Aliens landed on Earth!"]

# Convert to numbers
news_vector = vectorizer.transform(news)

# Predict
result = model.predict(news_vector)

print("Prediction:", result[0])
```

```python
# Check accuracy
accuracy = model.score(x_test, y_test)

print("Accuracy:", accuracy * 100)
```
